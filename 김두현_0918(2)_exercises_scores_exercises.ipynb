### Step 1. Import the necessary libraries

````python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
%matplotlib inline
````

### Step 2. Create the DataFrame that should look like the one below.

````python
raw_data = {
    'first_name': ['Jason', 'Molly', 'Tina', 'Jake', 'Amy'],
    'last_name': ['Miller', 'Jacobson', 'Ali', 'Milner', 'Cooze'],
    'female': [0, 1, 1, 0, 1],
    'age': [42, 52, 36, 24, 73],
    'preTestScore': [4, 24, 31, 2, 3],
    'postTestScore': [25, 94, 57, 62, 70]
}

df = pd.DataFrame(raw_data)
df
````

### Step 3. Create a Scatterplot of preTestScore and postTestScore, with the size of each point determined by age
#### Hint: Don't forget to place the labels

````python
plt.scatter(df['preTestScore'], df['postTestScore'], s=df['age'])

plt.title('preTestScore vs postTestScore (Size by Age)')
plt.xlabel('preTestScore')
plt.ylabel('postTestScore')
plt.show()
````

### Step 4. Create a Scatterplot of preTestScore and postTestScore.
### This time the size should be 4.5 times the postTestScore and the color determined by sex

````python
plt.scatter(df['preTestScore'], df['postTestScore'], s=df['postTestScore'] * 4.5, c=df['female'])

plt.title('preTestScore vs postTestScore')
plt.xlabel('preTestScore')
plt.ylabel('postTestScore')
plt.show()
````

### BONUS: Create your own question and answer it.

각 인물별 사전 점수와 사후 점수의 차이(향상도)는 얼마인가?

````python
# 점수 향상도 계산
df['score_diff'] = df['postTestScore'] - df['preTestScore']

# 막대 그래프 시각화
plt.bar(df['first_name'], df['score_diff'], color='teal')
plt.title('Score Improvement by Person')
plt.xlabel('Name')
plt.ylabel('Score Increase (post - pre)')
plt.show()
````
